```python
# --- Step 1: ダミーデータの準備 ---
grid_shape = (20, 30)
heatmap_data_A = np.random.rand(*grid_shape) * 20
heatmap_data_B = np.random.rand(*grid_shape) * 20
heatmap_data_A[0:10, 0:15] += 30
heatmap_data_B[10:20, 15:30] += 30


# --- Step 2: 差分データの計算 ---
diff_data = heatmap_data_A - heatmap_data_B


# --- Step 3: 差分データの可視化 (matplotlib ver.) ---
plt.figure(figsize=(12, 8))
plt.rcParams['font.family'] = 'sans-serif' # 日本語フォントを指定

# 色の範囲を0を中心に左右対称にするための計算
vmax = np.abs(diff_data).max() # 差分データの絶対値の最大値を取得
vmin = -vmax                   # 最大値にマイナスをつけた値を最小値とする

# plt.imshow() を使ってヒートマップを描画
im = plt.imshow(diff_data, cmap='RdBu_r', vmin=vmin, vmax=vmax)

# カラーバーを追加
plt.colorbar(im)

plt.title('滞在回数の差分ヒートマップ (条件A - 条件B)')
plt.xlabel('X座標グリッド')
plt.ylabel('Y座標グリッド')
plt.show()
```
